```python
df = spark.read.csv("/databricks-datasets/airlines/part-00000", header=True, inferSchema=True)
df.show(5)
```

```python
from pyspark.sql.functions import col

# Filter high-value orders
high_value = df.filter(col("total_spent") > 5000)  # 👈 Replace "80000" with actual column name

# Select key columns
result = high_value.select("Region", "total_spent", "Product")  # 👈 Replace with your actual column names

# Save inside Unity Catalog volume
result.write.mode("overwrite").parquet("/Volumes/workspace/default/sales_data/high_value_orders")

print("✅ Parquet file successfully saved in Unity Catalog volume!")
```
